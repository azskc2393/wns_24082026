```python
from delta.tables import DeltaTable
```

```python
spark.sql("""DROP TABLE IF EXISTS ecommerce.bronze.customers""")


(
  DeltaTable.createIfNotExists(spark)
    .tableName("ecommerce.bronze.customers")
    .addColumn("customer_id", "INT")
    .addColumn("first_name", "STRING")
    .addColumn("last_name", "STRING")
    .addColumn("email", "STRING")
    .addColumn("phone", "STRING")
    .addColumn("city", "STRING")
    .addColumn("state", "STRING")
    .addColumn("country", "STRING")
    .addColumn("customer_status", "STRING")
    .addColumn("created_at", "TIMESTAMP")
    .addColumn("updated_at", "TIMESTAMP")
    .execute()
)
```

```python
spark.sql("""DROP TABLE IF EXISTS ecommerce.bronze.inventory""")


(
  DeltaTable.createIfNotExists(spark)
    .tableName("ecommerce.bronze.inventory")
    .addColumn("inventory_date", "DATE")
    .addColumn("product_id", "INT")
    .addColumn("warehouse_id", "STRING")
    .addColumn("available_quantity", "INT")
    .addColumn("reserved_quantity", "INT")
    .addColumn("reorder_level", "INT")
    .execute()
)
```

```python
spark.sql("""DROP TABLE IF EXISTS ecommerce.bronze.products""")

(
  DeltaTable.createIfNotExists(spark)
    .tableName("ecommerce.bronze.products")
    .addColumn("product_id", "INT")
    .addColumn("product_name", "STRING")
    .addColumn("category", "STRING")
    .addColumn("brand", "STRING")
    .addColumn("price", "DOUBLE")
    .addColumn("cost", "DOUBLE")
    .addColumn("supplier_id", "STRING")
    .addColumn("product_status", "STRING")
    .addColumn("updated_at", "TIMESTAMP")
    .execute()
)
```
