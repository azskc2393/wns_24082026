```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
try:
    path = '/Volumes/ecommerce/quickstart/ecommerce_sandbox/ShopSphere_Ecommerce_Capstone_Data/adls/initial/products/products_2026-08-01.csv'

    products_df = spark.read.csv(path = path, header = True, inferSchema = True)

    products_df.write.saveAsTable('ecommerce.bronze.products')

except Exception as e:
    raise e
```

```python
%skip
try:
    PRODUCT_SCHEMA = StructType(
        [
            StructField("product_id", IntegerType()),
            StructField("product_name", StringType()),
            StructField("category", StringType()),
            StructField("subcategory", StringType()),
            StructField("brand", StringType()),
            StructField("price", DoubleType()),
            StructField("cost", DoubleType()),
            StructField("supplier_id", StringType()),
            StructField("product_status", StringType()),
            StructField("updated_at", TimestampType()),
        ]
    )

    path = "/Volumes/ecommerce/quickstart/ecommerce_sandbox/ShopSphere_Ecommerce_Capstone_Data/adls/initial/product_load/input/"

    input_df = spark.readStream.load(
        format="CSV", path=path, schema=PRODUCT_SCHEMA, header=True
    )

    input_df.writeStream.option(
        "checkpointLocation",
        "Volumes/ecommerce/quickstart/ecommerce_sandbox/ShopSphere_Ecommerce_Capstone_Data/adls/initial/product_load/checkpoint/",
    ).trigger(availableNow=True).outputMode("APPEND").toTable(
        "ecommerce.bronze.products"
    )

except Exception as e:
    raise e
```

```python
try:
    INVENTORY_SCHEMA = StructType(
        [
            StructField("inventory_date", DateType()),
            StructField("product_id", IntegerType()),
            StructField("warehouse_id", StringType()),
            StructField("available_quantity", IntegerType()),
            StructField("reserved_quantity", IntegerType()),
            StructField("reorder_level", IntegerType()),
        ]
    )

    path = "/Volumes/ecommerce/quickstart/ecommerce_sandbox/ShopSphere_Ecommerce_Capstone_Data/adls/initial/inventory_load/input/"

    input_df = spark.readStream.load(
        format="CSV", path=path, schema=INVENTORY_SCHEMA, header=True
    )

    input_df.writeStream.option(
        "checkpointLocation",
        "/Volumes/ecommerce/quickstart/ecommerce_sandbox/ShopSphere_Ecommerce_Capstone_Data/adls/initial/inventory_load/checkpoint/",
    ).trigger(availableNow=True).outputMode("APPEND").toTable(
        "ecommerce.bronze.inventory"
    )

except Exception as e:
    raise e
```

```python
try:
    CLICKSTREAM_SCHEMA = StructType(
        [
            StructField("customer_id", DoubleType()),
            StructField("device_type", StringType()),
            StructField("event_id", LongType()),
            StructField("event_time", StringType()),
            StructField("event_type", StringType()),
            StructField("page", StringType()),
            StructField("product_id", LongType()),
            StructField("session_id", StringType()),
        ]
    )

    path = "/Volumes/ecommerce/quickstart/ecommerce_sandbox/ShopSphere_Ecommerce_Capstone_Data/adls/initial/clickstream_load/input/"

    input_df = spark.readStream.load(
        format="JSON", path=path, schema=CLICKSTREAM_SCHEMA
    )

    input_df.writeStream.option(
        "checkpointLocation",
        "/Volumes/ecommerce/quickstart/ecommerce_sandbox/ShopSphere_Ecommerce_Capstone_Data/adls/initial/clickstream_load/checkpoint/",
    ).trigger(availableNow=True).outputMode("APPEND").toTable(
        "ecommerce.bronze.clickstream"
    )

except Exception as e:
    raise e
```
