## DATABRICKS TRAINING SESSIONS
<br>**DATE: 2026-09-01**
<br>**TOPIC: Streaming Files**

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
CRIME_SCHEMA = StructType(
    [
        StructField("code", StringType()),
        StructField("region", StringType()),
        StructField("category", StringType()),
    ]
)

input_df = spark.readStream.load(
    format="CSV", path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/crime_data/input/", schema=CRIME_SCHEMA
)

result_df = input_df.groupBy("region").count()

result_df.writeStream.option(
    "checkpointLocation",
    "/Volumes/wns24082026/quickstart_schema/sandbox/datasets/crime_data/checkpoint/",
).trigger(availableNow=True).outputMode("COMPLETE").toTable(
    "wns24082026.quickstart_schema.crime"
)
```
