# Adding Custom Schema for csv files

```python
from pyspark.sql.types import *

USER_SCHEMA = StructType(
    [
    StructField("id", IntegerType()),
    StructField("age", IntegerType()),
    StructField("gender", StringType()),
    StructField("designation", StringType()),
    StructField("salary", IntegerType())
]
    )

df = spark.read.csv("/Volumes/workspace/default/sandbox/dataset/sample_dataset/users_001.csv", header=True, schema=USER_SCHEMA)
df.display()

### WHENEVER WE ATTACH A MISMATCH SCHEMA, WE GET NULL VALUES IN THE COLUMN
```

# Handling Bad Records

1. Permissive Mode
2. DropMalFormed
3. FailFast


```python
df_ff = spark.read.json(
    "/Volumes/workspace/default/sandbox/dataset/access_logs.json", mode="FAILFAST"
)

df_ff_1 = spark.read.option("columnNameOfCorruptRecord", "bad_records").json(
    "/Volumes/workspace/default/sandbox/dataset/access_logs.json", mode="FAILFAST"
).display()


df_dmf = spark.read.json(
    "/Volumes/workspace/default/sandbox/dataset/access_logs.json", mode="DROPMALFORMED"
)


df_per = spark.read.json(
    "/Volumes/workspace/default/sandbox/dataset/access_logs.json", mode="PERMISSIVE"
)

df_per_col_up = spark.read.option("columnNameOfCorruptRecord", "bad_records").json(
    "/Volumes/workspace/default/sandbox/dataset/access_logs.json", mode="PERMISSIVE"
)

df_per.display()
df_per_col_up.display()
df_dmf.display()
df_ff.display()
```

# Output Mode
1. ErrorExist
2. Overwrite
3. Append
4. Ignore

```python
path = "/Volumes/workspace/default/sandbox/dataset/sample_dataset/users_001.csv"
df = dfr.csv(path, header=True, inferSchema=True,sep = ",")

# ERROR EXISTS
df.write.save(format="parquet", path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/output/output_parquet")


# OVERWRITE
df.write.save(
    format="json",
    mode="overwrite",
    path="/Volumes/workspace/default/sandbox/output.json",
)

# APPEND
df.write.save(
    format="json", mode="append", path="/Volumes/workspace/default/sandbox/output.json"
)

# IGNORE
df.write.save(
    format="json", mode="ignore", path="/Volumes/workspace/default/sandbox/output.json"
)

df_json = spark.read.json("/Volumes/workspace/default/sandbox/output.json")

df_json.count()
```
