#Prerequisite: Load Employee csv

# SELECT OPERATIONS

#Benefts of using col is that we can assign an alias name in the select itself, we cannot have it using ""

```python
df.select("id", "name").show()

from pyspark.sql.functions import col

df.select(col("id").alias('upd_id'), col("name")).display() # Recommended approach
```
