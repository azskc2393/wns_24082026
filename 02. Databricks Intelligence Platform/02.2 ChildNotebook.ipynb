```python
from pyspark.sql.functions import when, col, count

dbutils.widgets.dropdown(
    name="exp_level",
    defaultValue="Invalid experience",
    choices=["Senior", "Mid Level", "Junior", "Invalid experience"],
    label="Experience",
)

path = "/Volumes/wns24082026/quickstart_schema/sandbox/datasets/employee.csv"
employee = spark.read.csv(path, header=True, inferSchema=True, sep="|", quote="'")


df_exp = (
    employee.withColumn(
        "exp_level",
        when(col("exp") >= 10, "Senior")
        .when(col("exp") >= 5, "Mid Level")
        .when(col("exp") >= 0, "Junior")
        .otherwise("Invalid experience"),
    )
    .select(col("name"), col("exp"), col("exp_level"))
    .groupBy(col("exp_level"))
    .agg(count(col("exp")).alias("employeecount"))
)

explvl = dbutils.widgets.get("exp_level")
df_exp.filter(col("exp_level") == explvl).display()
```
