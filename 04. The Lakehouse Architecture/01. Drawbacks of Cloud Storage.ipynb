#Prerequisites
**Load 'users_001.csv' into DataFrame**

```python
path = '/Volumes/wns24082026/quickstart_schema/sandbox/datasets/user_dataset/users_001.csv'
df = spark.read.csv(path = path, header = True, inferSchema = True)
df.display()
```

#Write as Parquet

```python
df.write.save(
    format="Parquet",
    mode='overwrite',
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/output",
)
```

#Read as parquet

```python
df = spark.read.format("parquet").load(
    "/Volumes/wns24082026/quickstart_schema/sandbox/datasets/output"
)
df.display()
```

#Execute sql query on dataframe

```python
df.createOrReplaceTempView('users_vw')
```

# [Disadvantage] Updates are not allowed

```python
%sql
UPDATE
  users_vw
SET
  country = 'Bharat'
WHERE
  country = 'India'
```

# [Disadvantage] Roll back is not allowed once overwritten

```python
from pyspark.sql.functions import *

df.filter(col("country") == "India").write.save(
    format="parquet",
    mode="overwrite",
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/output",
)
```

```python
spark.read.format("parquet").load(
    "/Volumes/wns24082026/quickstart_schema/sandbox/datasets/output"
).display()
```
