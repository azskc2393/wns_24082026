```python
from pyspark.sql.functions import *
path = "/Volumes/wns24082026/quickstart_schema/sandbox/datasets/user_dataset/users_011_datatype_mismatch_dob.csv"

df = spark.read.csv(path, header=True, inferSchema=True)

df = df.withColumn('dob', to_date(col('dob')))

df.write.option("mergeSchema", True).saveAsTable(
    "wns24082026.quickstart_schema.users", mode="OVERWRITE"
)

## Reason for failure: cannot convert the dob column from long to date- default value is date
```

```python
path = "/Volumes/wns24082026/quickstart_schema/sandbox/datasets/user_dataset/users_013_less_columns_with_different_column_names.csv"

df = spark.read.csv(path, header=True, inferSchema=True).select(col('user_id').alias('id'),
                                                                col('full_name').alias('name'))

df.write.option("mergeSchema", True).saveAsTable(
    "wns24082026.quickstart_schema.users", mode="APPEND"
)
```
